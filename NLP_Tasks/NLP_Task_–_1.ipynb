# **Task 2:Build Advanced Preprocessing Function**


```python
import re
from collections import Counter
def preprocess_text(text):

    # Safety check
    if text is None or not isinstance(text, str):
        return [], ""

    # Lowercase first
    text = text.lower()

    # Remove links (http, https, www)
    text = re.sub(r'(https?://\S+|www\.\S+)', '', text)

    # Remove emails
    text = re.sub(r'\S+@\S+', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove emojis / special characters (keep letters + spaces)
    text = re.sub(r'[^a-z\s]', ' ', text)

    # Fix repeated characters (but smarter)
    text = re.sub(r'(.)\1{2,}', r'\1', text)

    # Clean extra spaces
    text = " ".join(text.split())

    # Tokenization
    words = text.split()

    # Remove short words (keep "no", "not")
    final_tokens = []
    for w in words:
        if len(w) > 2 or w in ["no", "not"]:
            final_tokens.append(w)

    # Join back
    cleaned_sentence = " ".join(final_tokens)

    return final_tokens, cleaned_sentence
```

# **Task 3: Stress Testing**


```python
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]
all_tokens = []

for s in sentences:
    tokens, clean = preprocess_text(s)
    all_tokens.extend(tokens)

    print("Original:", s)
    print("Tokens:", tokens)
    print("Cleaned:", clean)
    print("-" * 50)
```

# **Task 4: Token Analytics**


```python
for s in sentences:
    tokens, _ = preprocess_text(s)

    total = len(tokens)
    unique = len(set(tokens))

    avg = sum(len(t) for t in tokens) / total if total else 0

    print("Sentence:", s)
    print("Total:", total)
    print("Unique:", unique)
    print("Avg Length:", round(avg, 2))
    print("-" * 40)
```

# **Task 5: Frequency Analysis**


```python
freq = Counter(all_tokens)

print("Top 10:", freq.most_common(10))
print("Least 5:", freq.most_common()[-5:])
```

# **Task 6: Build Full Pipeline**


```python

def full_pipeline(text_list):

    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean_text = preprocess_text(text)

        all_tokens.extend(tokens)
        clean_sentences.append(clean_text)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }


result = full_pipeline(sentences)

print("All Tokens:", result["tokens"])
print("\nClean Sentences:", result["clean_sentences"])
```
